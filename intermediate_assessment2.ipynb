```python
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]
    return outliers
```

```python
for col in numerical_columns:
    outliers = detect_outliers_iqr(df, col)
    print(col, ":", len(outliers), "outliers")
```

```python
correlation = df.corr(numeric_only=True)["chol"].sort_values(ascending=False)
print(correlation)
```

```python
selected_reg_features = correlation[
    correlation.abs() >= 0.1
].index.tolist()

selected_reg_features.remove("chol")

print("Selected Regression Features:")
print(selected_reg_features)
```

```python
X_reg = df[selected_reg_features]
y_reg = df["chol"]
```

```python
X_reg = pd.get_dummies(
    X_reg,
    columns=["restecg"],
    drop_first=True,
    dtype=int
)
```

```python
from sklearn.model_selection import train_test_split
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler
scaler_reg = StandardScaler()
X_reg_train_scaled = scaler_reg.fit_transform(X_reg_train)
X_reg_test_scaled = scaler_reg.transform(X_reg_test)
```

```python
linear_model = LinearRegression()
linear_model.fit(
    X_reg_train_scaled,
    y_reg_train
)
linear_pred = linear_model.predict(X_reg_test_scaled)
```

```python
svm_model = SVR()
svm_model.fit(
    X_reg_train_scaled,
    y_reg_train
)
svm_pred = svm_model.predict(X_reg_test_scaled)
```

```python
rf_reg_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
rf_reg_model.fit(
    X_reg_train,
    y_reg_train
)
rf_reg_pred = rf_reg_model.predict(X_reg_test)
```

```python
def evaluate_regression(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    return mae, mse, r2
```

```python
linear_results = evaluate_regression(
    y_reg_test,
    linear_pred
)
svm_results = evaluate_regression(
    y_reg_test,
    svm_pred
)
rf_reg_results = evaluate_regression(
    y_reg_test,
    rf_reg_pred
)
```

```python
regression_results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "SVM",
        "Random Forest"
    ],
    "MAE": [
        linear_results[0],
        svm_results[0],
        rf_reg_results[0]
    ],
    "MSE": [
        linear_results[1],
        svm_results[1],
        rf_reg_results[1]
    ],
    "R2": [
        linear_results[2],
        svm_results[2],
        rf_reg_results[2]
    ]
})
regression_results
```

```python
best_regression_model = regression_results.loc[
    regression_results["R2"].idxmax()
]
print("Best Regression Model:")
print(best_regression_model)
```

```python
correlation_target = df.corr(numeric_only=True)["target"].sort_values(
    ascending=False
)
print(correlation_target)
```

```python
selected_clf_features = correlation_target[
    correlation_target.abs() >= 0.1
].index.tolist()

selected_clf_features.remove("target")

print("Selected Classification Features:")
print(selected_clf_features)
```

```python
X_clf = df[selected_clf_features]
y_clf = df["target"]
```

```python
X_clf = pd.get_dummies(
    X_clf,
    columns=["cp", "restecg", "thal"],
    drop_first=True,
    dtype=int
)
```

```python
X_clf_train, X_clf_test, y_clf_train, y_clf_test = train_test_split(
    X_clf,
    y_clf,
    test_size=0.2,
    random_state=42,
    stratify=y_clf
)
```

```python
scaler_clf = StandardScaler()
X_clf_train_scaled = scaler_clf.fit_transform(X_clf_train)
X_clf_test_scaled = scaler_clf.transform(X_clf_test)
```

```python
logistic_model = LogisticRegression(
    random_state=42,
    max_iter=1000
)
logistic_model.fit(
    X_clf_train_scaled,
    y_clf_train
)
logistic_pred = logistic_model.predict(X_clf_test_scaled)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_clf_test,
    logistic_pred
)

plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(
    X_clf_train_scaled,
    y_clf_train
)
knn_pred = knn_model.predict(X_clf_test_scaled)
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_clf_test,
    knn_pred
)

plt.title("KNN - Confusion Matrix")
plt.show()
```

```python
rf_clf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
rf_clf_model.fit(
    X_clf_train,
    y_clf_train
)
rf_clf_pred = rf_clf_model.predict(X_clf_test)
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_clf_test,
    rf_clf_pred
)

plt.title("Random Forest - Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

```python
def evaluate_classification(y_true, y_pred):
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    return accuracy, precision, recall, f1
```

```python
logistic_results = evaluate_classification(
    y_clf_test,
    logistic_pred
)
knn_results = evaluate_classification(
    y_clf_test,
    knn_pred
)
rf_clf_results = evaluate_classification(
    y_clf_test,
    rf_clf_pred
)
```

```python
classification_results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "KNN",
        "Random Forest"
    ],
    "Accuracy": [
        logistic_results[0],
        knn_results[0],
        rf_clf_results[0]
    ],
    "Precision": [
        logistic_results[1],
        knn_results[1],
        rf_clf_results[1]
    ],
    "Recall": [
        logistic_results[2],
        knn_results[2],
        rf_clf_results[2]
    ],
    "F1-Score": [
        logistic_results[3],
        knn_results[3],
        rf_clf_results[3]
    ]
})

classification_results
```

```python
best_classification_model = classification_results.loc[
    classification_results["Recall"].idxmax()
]

print("Best Classification Model:")
print(best_classification_model)
```
